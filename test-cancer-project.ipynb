```python
import requests
import json
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

```python
scoring_uri = "PASTE_YOUR_SCORING_URI_FROM_THE_LAST_STEP"

# --- 2. Get a raw data sample (the same one) ---
data = load_breast_cancer()
raw_data_sample = data.data[0] # Get the first row
true_value = data.target[0]   # True value is 0 (Malignant)

# --- 3. Format as a pandas split ---
payload_df = pd.DataFrame([raw_data_sample], columns=data.feature_names)
payload_dict = {"dataframe_split": payload_df.to_dict(orient="split")}
input_json = json.dumps(payload_dict)

# --- 4. Send the request ---
headers = {'Content-Type': 'application/json'}
resp = requests.post(scoring_uri, input_json, headers=headers)

# --- 5. Print results ---
print(f"Raw data sent:\n{raw_data_sample}\n")
print(f"True value: {true_value} (0=Malignant, 1=Benign)\n")
print(f"API Response:\n{resp.text}")
```
